```python
import os 
import json 
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
load_dotenv(override=True)

openai_api_key = os.getenv('OPENAI_API_KEY')
if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")

openai = OpenAI(base_url="http://localhost:11434/v1",api_key="ollama")
MODEL = "llama3.1:latest"
```

```python
system_message="""
You are a helpful AI assistant for an airline company. You can help users with flight bookings, cancellations
, and provide information about flight schedules, baggage policies, and other travel-related inquiries.
You answer everything with one line in concise manner. You are  allowed to answer anything that is not related to the airline company but in short not more than 2 lines .
"""
```

```python
def chat(message, history):
    history = [{"role":h["role"], "content":h["content"]} for h in history]
    messages = [{"role": "system", "content": system_message}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model=MODEL, messages=messages)
    return response.choices[0].message.content

gr.ChatInterface(fn=chat).launch()
```

```python
ticket_prices={
    "london":"Rs 2,00,000",
    "paris":"Rs 1,50,000",
    "new york":"Rs 2,50,000",
    "abu dhabi":"Rs 1,80,000",
}
def get_ticket_price(destination):
    print(f"Tool called for city: {destination}")
    price=ticket_prices.get(destination.lower(),"unknown ticket price")
    return f"The price of a ticket to {destination} is {price}."
```
